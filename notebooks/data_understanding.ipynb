```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import openpyxl
```

```python
FILE_PATH = r"E:\SEM5\DPTLM\DTPLM_Project\college_predictor\Data\MHTCET_Engineering_CutOffs_2022-2025.csv"

df = pd.read_csv(FILE_PATH)

print("Dataset loaded successfully.")
print("=" * 70)
```

```python
print("\nDATASET SHAPE")
print("-" * 70)
print("Rows    :", df.shape[0])
print("Columns :", df.shape[1])

print("\nCOLUMN NAMES")
print("-" * 70)
for i, col in enumerate(df.columns, 1):
    print(i, ":", col)

print("\nFIRST 10 RECORDS")
print("-" * 70)
display(df.head(10))

print("\nLAST 10 RECORDS")
print("-" * 70)
display(df.tail(10))
```

```python
print("\nDATA TYPES")
print("-" * 70)
display(
    pd.DataFrame({
        "Column": df.columns,
        "Data_Type": df.dtypes.astype(str).values,
        "Unique_Values": [df[c].nunique() for c in df.columns],
        "Missing_Values": [df[c].isna().sum() for c in df.columns]
    })
)
```

```python
print("\nMISSING VALUE ANALYSIS")
print("-" * 70)

missing = pd.DataFrame({
    "Column": df.columns,
    "Missing_Count": df.isna().sum(),
    "Missing_Percentage": (df.isna().mean() * 100).round(4)
})

display(missing)

print("\nTotal missing cells:", df.isna().sum().sum())
```

```python
print("\nDUPLICATE ANALYSIS")
print("-" * 70)

exact_duplicates = df.duplicated().sum()

print("Exact duplicate rows:", exact_duplicates)

if exact_duplicates > 0:
    display(df[df.duplicated(keep=False)].sort_values(df.columns.tolist()).head(50))
```

# YEAR ANALYSIS

```python
print("\nYEAR ANALYSIS")
print("-" * 70)

year_counts = (
    df["Year"]
    .value_counts()
    .sort_index()
    .rename_axis("Year")
    .reset_index(name="Records")
)

display(year_counts)

print("\nAcademic Year mapping:")
display(
    df[["Year", "AcademicYear"]]
    .drop_duplicates()
    .sort_values("Year")
)

plt.figure(figsize=(8, 5))
sns.barplot(data=year_counts, x="Year", y="Records")
plt.title("Records by Year")
plt.xlabel("Year")
plt.ylabel("Number of Records")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
print("\nCAP ROUND ANALYSIS")
print("-" * 70)

cap_counts = (
    df["CAPRound"]
    .value_counts()
    .sort_index()
    .rename_axis("CAPRound")
    .reset_index(name="Records")
)

display(cap_counts)

print("\nYear × CAP Round")
display(
    pd.crosstab(
        df["Year"],
        df["CAPRound"]
    )
)

plt.figure(figsize=(8, 5))
sns.countplot(data=df, x="CAPRound", order=sorted(df["CAPRound"].unique()))
plt.title("Records by CAP Round")
plt.xlabel("CAP Round")
plt.ylabel("Number of Records")
plt.tight_layout()
plt.show()
```

# COLLEGE ANALYSIS

```python
print("\nCOLLEGE ANALYSIS")
print("-" * 70)

print("Unique College Codes :", df["CollegeCode"].nunique())
print("Unique College Names :", df["CollegeName"].nunique())

college_records = (
    df["CollegeName"]
    .value_counts()
    .reset_index()
)

college_records.columns = ["CollegeName", "Records"]

display(college_records.head(30))

print("\nColleges per year:")
display(
    df.groupby("Year")["CollegeCode"]
    .nunique()
    .reset_index(name="Unique_Colleges")
)
```

## COLLEGE CODE CONSISTENCY CHECK

```python
print("\nCOLLEGE CODE / NAME CONSISTENCY")
print("-" * 70)

college_code_name_count = (
    df.groupby("CollegeCode")["CollegeName"]
    .nunique()
    .sort_values(ascending=False)
)

college_conflicts = college_code_name_count[
    college_code_name_count > 1
]

print(
    "College codes associated with multiple names:",
    len(college_conflicts)
)

if len(college_conflicts) > 0:

    college_conflict_table = (
        df[df["CollegeCode"].isin(college_conflicts.index)]
        [["Year", "CollegeCode", "CollegeName"]]
        .drop_duplicates()
        .sort_values(["CollegeCode", "Year", "CollegeName"])
    )

    display(college_conflict_table)
```

## BRANCH ANALYSIS

```python
print("\nBRANCH ANALYSIS")
print("-" * 70)

print("Unique Branch Codes :", df["BranchCode"].nunique())
print("Unique Branch Names :", df["BranchName"].nunique())

branch_records = (
    df["BranchName"]
    .value_counts()
    .reset_index()
)

branch_records.columns = ["BranchName", "Records"]

display(branch_records.head(50))

print("\nBranches per year:")
display(
    df.groupby("Year")["BranchCode"]
    .nunique()
    .reset_index(name="Unique_Branches")
)
```

## BRANCH CODE CONSISTENCY ANALYSIS

```python
print("\nBRANCH CODE / NAME CONSISTENCY")
print("-" * 70)

branch_code_name_count = (
    df.groupby("BranchCode")["BranchName"]
    .nunique()
    .sort_values(ascending=False)
)

branch_conflicts = branch_code_name_count[
    branch_code_name_count > 1
]

print(
    "Branch codes associated with multiple names:",
    len(branch_conflicts)
)

if len(branch_conflicts) > 0:

    branch_conflict_table = (
        df[df["BranchCode"].isin(branch_conflicts.index)]
        [["Year", "BranchCode", "BranchName"]]
        .drop_duplicates()
        .sort_values(["BranchCode", "Year", "BranchName"])
    )

    display(branch_conflict_table)
```

# LOGICAL DUPLICATE ANALYSIS

```python
print("\nLOGICAL DUPLICATE ANALYSIS")
print("-" * 70)

logical_key = [
    "Year",
    "CAPRound",
    "CollegeCode",
    "BranchCode",
    "Category",
    "Stage"
]

logical_counts = (
    df.groupby(
        logical_key,
        dropna=False
    )
    .size()
    .reset_index(name="RecordCount")
)

multi_record_groups = logical_counts[
    logical_counts["RecordCount"] > 1
].sort_values(
    "RecordCount",
    ascending=False
)

print(
    "Logical groups:",
    len(logical_counts)
)

print(
    "Groups containing multiple records:",
    len(multi_record_groups)
)

display(
    multi_record_groups.head(50)
)
```

# CUTOFF CANDIDATE AGGREGATION

```python
print("\nCUTOFF CANDIDATE ANALYSIS")
print("-" * 70)

cutoff_key = [
    "Year",
    "CAPRound",
    "CollegeCode",
    "CollegeName",
    "BranchCode",
    "BranchName",
    "Category",
    "Stage"
]

cutoff_candidates = (
    df.groupby(
        cutoff_key,
        dropna=False
    )
    .agg(
        RecordCount=("Percentile", "size"),

        MinRank=("Rank", "min"),
        MaxRank=("Rank", "max"),

        MinPercentile=("Percentile", "min"),
        MaxPercentile=("Percentile", "max"),

        MedianRank=("Rank", "median"),
        MedianPercentile=("Percentile", "median")
    )
    .reset_index()
)

display(
    cutoff_candidates.head(50)
)
```

# NEW COLLEGE BRANCH

```python
print("\nNEW COLLEGE/BRANCH COMBINATIONS")
print("-" * 70)

for year in sorted(df["Year"].unique()):

    current = set(
        zip(
            df.loc[df["Year"] == year, "CollegeCode"],
            df.loc[df["Year"] == year, "BranchCode"]
        )
    )

    previous = set()

    for previous_year in sorted(df["Year"].unique()):

        if previous_year < year:

            previous.update(
                zip(
                    df.loc[
                        df["Year"] == previous_year,
                        "CollegeCode"
                    ],
                    df.loc[
                        df["Year"] == previous_year,
                        "BranchCode"
                    ]
                )
            )

    new_combinations = current - previous

    print(
        year,
        "new college/branch combinations:",
        len(new_combinations)
    )
```

# CHECK FOR TEXT WHITESPACE

```python
print("\nTEXT WHITESPACE CHECK")
print("-" * 70)

text_columns = df.select_dtypes(
    include=["object"]
).columns

whitespace_report = []

for col in text_columns:

    leading = df[col].astype(str).str.startswith(" ").sum()
    trailing = df[col].astype(str).str.endswith(" ").sum()

    whitespace_report.append({
        "Column": col,
        "LeadingSpaces": leading,
        "TrailingSpaces": trailing
    })

display(
    pd.DataFrame(whitespace_report)
)
```

## NEWLINE/ TAB CHARACTERS

```python
print("\nNEWLINE / TAB CHARACTER CHECK")
print("-" * 70)

special_character_report = []

for col in text_columns:

    newline_count = df[col].astype(str).str.contains(
        "\n",
        regex=False
    ).sum()

    tab_count = df[col].astype(str).str.contains(
        "\t",
        regex=False
    ).sum()

    special_character_report.append({
        "Column": col,
        "NewlineRecords": newline_count,
        "TabRecords": tab_count
    })

display(
    pd.DataFrame(special_character_report)
)
```
